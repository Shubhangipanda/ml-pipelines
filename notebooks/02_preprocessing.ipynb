```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# -----------------------------
# 1. Load Data
# -----------------------------
df = pd.read_csv("/Users/shubhangipanda/Documents/Projects/ml-pipelines/data/raw/Churn.csv")
```

```python
print("Shape before cleaning:", df.shape)
df.head()
```

```python
# 2. Handle Missing Values
# -----------------------------
# "TotalCharges" is numeric but stored as object (some empty strings)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
# 3. Drop Irrelevant Columns
# -----------------------------
# 'customerID' is just an identifier
df.drop("customerID", axis=1, inplace=True)
```

```python
# 4. Encode Categorical Variables
# -----------------------------
# Binary columns: Yes/No → 1/0
binary_cols = [col for col in df.columns if df[col].nunique() == 2]
```

```python
for col in binary_cols:
    df[col] = df[col].map({"Yes": 1, "No": 0})
```

```python
# Multi-category columns: use LabelEncoder
cat_cols = df.select_dtypes(include=["object"]).columns.tolist()
```

```python
print(cat_cols)
```

```python

le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df.head()
```

```python
# 5. Scale Numerical Features
# -----------------------------
num_cols = ["tenure", "MonthlyCharges", "TotalCharges"]
```

```python
scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
# 6. Save Cleaned Data
# -----------------------------
df.to_csv("/Users/shubhangipanda/Documents/Projects/ml-pipelines/data/processed/Churn.csv", index=False)
```
